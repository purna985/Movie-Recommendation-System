```python
import numpy as np
import pandas as pd
```

```python
movies=pd.read_csv('tmdb_5000_movies.csv')
credits=pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies=movies.merge(credits,on='title')
```

```python
movies=movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.info()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies.iloc[0].genres
```

```python
type(movies.iloc[0].genres)
```

```python
import ast
def convert(obj):
  L=[]
  for i in ast.literal_eval(obj):
    L.append(i['name'])
  return L
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
import ast
def convert3(obj):
  L=[]
  counter=0
  for i in ast.literal_eval(obj):
    if counter!=3:
     L.append(i['name'])
     counter+=1
    else:
        break
  return L
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
import ast
def fetch_director(obj):
  L=[]
  for i in ast.literal_eval(obj):
    if i['job']=='Director':
      L.append(i['name'])
      break
  return L
```

```python
movies['crew']=movies['crew'].apply(fetch_director)
```

```python
movies['overview'][0]
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
def merge(L):
 L1=[]
 for i in L:
     L1.append(i.replace(" ",""))
 return L1
```

```python
movies['genres']=movies['genres'].apply(merge)
movies['keywords']=movies['keywords'].apply(merge)
movies['cast']=movies['cast'].apply(merge)
movies['crew']=movies['crew'].apply(merge)
```

```python
movies.head()
```

```python
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
movies.head()
```

```python
newdf=movies[['movie_id','title','tags']]
```

```python
newdf.head()
```

```python
type(newdf['tags'][0])
```

```python
newdf['tags'] = newdf['tags'].apply(lambda x:" ".join(x))
```

```python
newdf.head()
```

```python
newdf['tags'] = newdf['tags'].apply(lambda x:x.lower())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors=cv.fit_transform(newdf['tags'] ).toarray()
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
 y=[]
 for i in text.split():
    y.append(ps.stem(i))
 return " ".join(y)   
```

```python
newdf['tags']=newdf['tags'].apply(stem)
```

```python
newdf['tags']
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
def recommend(movie):
 movie_index=newdf[newdf['title']==movie].index[0]
 distances=similarity[movie_index]
 movies_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
 for i in movies_list:
   print(newdf.iloc[i[0]].title)
```

```python
recommend('Batman Begins')
```
